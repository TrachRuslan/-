```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
X = np.array([
    [0],
    [1],
    [2],
    [3]
])

Y = np.array([
    [0],
    [1],
    [2],
    [3]
])
```

```python
model = Sequential()

model.add(Dense(
    1,
    input_dim=1
))
```

```python
model.compile(
    optimizer='adam',
    loss='mean_squared_error'
)
```

```python
model.fit(
    X,
    Y,
    epochs=100,
    verbose=1
)
```

```python
print(model.predict(np.array([[5]])))
```
