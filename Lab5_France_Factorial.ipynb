```python
def factorial(n):

    if not isinstance(n, int):
        return "Помилка: потрібно ввести ціле число."

    if n < 0:
        return "Помилка: факторіал існує тільки для невід'ємних чисел."

    result = 1

    for i in range(1, n + 1):
        result *= i

    return f"Факторіал числа {n} = {result}"
```

```python
print(factorial(5))
print(factorial(7))
print(factorial(10))
```

```python
import matplotlib.pyplot as plt

numbers = [1, 2, 3, 4, 5, 6]
factorials = []

for n in numbers:
    result = 1

    for i in range(1, n + 1):
        result *= i

    factorials.append(result)

plt.figure(figsize=(8,5))
plt.plot(numbers, factorials, marker='o')

plt.title("Графік факторіала")
plt.xlabel("Число")
plt.ylabel("Факторіал")

plt.grid(True)

plt.show()
```
